### Task 1 - Teoría

**1. Investigar el algoritmo AC-3 y su relación con el algoritmo de backtracking search**

El algoritmo AC-3 (Arc Consistency Algorithm 3) es un método utilizado para hacer que un problema de satisfacción de restricciones (CSP) sea más fácil de resolver. Funciona eliminando valores de los dominios de las variables que no cumplen con las restricciones binarias, asegurando que cada valor de una variable tenga al menos un valor compatible en las variables relacionadas.

Relación con el backtracking search:
AC-3 se puede usar antes o durante la búsqueda con backtracking para reducir el tamaño del espacio de búsqueda. Al aplicar AC-3, se eliminan combinaciones inválidas anticipadamente, lo que ayuda a detectar inconsistencias más temprano y disminuye el número de retrocesos (backtracks). Esto hace que el algoritmo de búsqueda sea más eficiente.

**2. Defina en sus propias palabras el término “Arc Consistency”**
Una variable en un problema de restricciones es arc-consistente si, para cada valor en su dominio, existe al menos un valor en el dominio de la variable relacionada que satisface la restricción entre ambas. Es decir, no hay valores “sin futuro”. Arc Consistency ayuda a reducir el dominio de las variables sin eliminar soluciones válidas, facilitando la resolución del problema.
