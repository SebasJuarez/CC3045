### Task 1 - Teoría

**1. En un juego de suma cero para dos jugadores, ¿cómo funciona el algoritmo minimax para determinar la estrategia óptima para cada jugador?**

El algoritmo Minimax se usa en juegos de suma cero para dos jugadores, por ejemplo el ajedrez, con i"nformación perfecta". La idea es encontrar la mejor estrategia para un jugador, asumiendo que el oponente jugará de cierta manera, o lo mejor posible.

**¿Puede explicarnos el concepto de "valor minimax" y su importancia en este contexto?**

El valor minimax de un estado del juego es el resultado que se puede esperar si ambos jugadores siguen las estrategias esperadas, por ejemplo:
- Un jugador (al que podemos llamar Maximizador) intenta obtener la mejor puntuación posible.
- El otro jugador (el Minimizador en este caso) intenta reducir la puntuación del primero.

El algoritmo recorre el árbol de juego desde las hojas hasta la raíz, asignando valores a cada nodo de acuerdo con si pertenece al jugador que maximiza o al que minimiza. La estrategia que se espera para el jugador "maximizador" es elegir el movimiento con el valor mas alto del resultado de minimax.

El valor minimax es muy importante porque garantiza la mejor jugada posible contra un oponente que juega bien o segun lo esperado. Ayuda a leer los movimientos del rival y a tomar decisiones basadas en evaluaciones de las respuestas para el futuro del juego.



**2. Compare y contraste el algoritmo minimax con la poda alfa-beta. ¿Cómo mejora la poda alfa-beta la eficiencia del algoritmo minimax, particularmente en árboles de caza grandes? Proporcione un ejemplo para ilustrar la diferencia en la complejidad computacional entre la poda minimax y alfa-beta.**

El algoritmo Minimax tiene una complejidad O(b^d), donde:
- b es el número de movimientos posibles por turno.
- d es la profundidad máxima del árbol.
Cuando los juegos tienen árboles de búsqueda muy grandes, el tiempo de ejecución puede volverse muy complejo.

**¿Cómo mejora la poda alfa-beta la eficiencia del minimax?**

La poda alfa-beta reduce el número de nodos que se evaluan sin que esto afecte al resultado final. La idea es que si se evaluaron algunos movimientos y hay otros peores, esos no evaluan por que claramente es peor

**Ejemplo:**

Sin poda alfa-beta, Minimax analiza b^d nodos. Con poda alfa-beta, el peor caso sigue siendo O(b^d), pero en el mejor caso cambia a O(b^(d/2)), reduciendo mucho el tiempo de ejecucion y mejorando la eficienca, ademas de que no afecta al resultado final.

Viendo esto en un juego, si un árbol de juego sin poda tiene b = 10 y d = 6, se hace una evaluación de 10^6 = 1,000,000 nodos. Con poda alfa-beta, podría reducirse a 10^3 = 1,000 nodos, acelerando todo el proceso de por medio.


**3. ¿Cuál es el papel de expectiminimax en juegos con incertidumbre, como aquellos que involucran nodos de azar o información oculta? ¿En qué se diferencia el expectiminimax del minimax en el manejo de resultados probabilísticos y cuáles son los desafíos clave que aborda?**

El algoritmo Expectiminimax es una "versión" del Minimax para juegos con ciertos elementos que pueden ser aleatorios, como juegos de cartas.

**Diferencia clave entre Minimax y Expectiminimax**

Minimax funciona suponiendo que los jugadores tienen el control total del juego y toman decisiones especificas, a veces siendo estas las mejores. Expectiminimax maneja nodos de azar o random, lo que introduce valores con probabilidades en la evaluación de acciones.

En la manera que se manejan los resultados de probabilidades, nn lugar de solo evaluar nodos MAX y MIN, Expectiminimax introduce nodos de "Expectación", donde:

- Se calcula el valor esperado en nodos con eventos aleatorios.
- Se pondera cada posible resultado según su probabilidad.
- Se propaga el valor esperado hacia arriba en el árbol de decisión.

Ejemplo en un juego de dados:

Si una jugada depende de lanzar un dado, el nodo de azar promediará los valores de cada posible resultado multiplicado por su probabilidad (1/6 por cada cara del dado).

**Desafíos clave**

- Explosión combinatoria: Al agregar nodos de azar, el tamaño del árbol de búsqueda aumenta significativamente.
- Estimación de probabilidades: Requiere modelar probabilidades con precisión.
- Optimización de búsqueda: Se pueden usar técnicas como la poda en juegos con azar (poda alfa-beta no se aplica directamente).
