### Task 1 - Teoría

**1. ¿Qué es el temporal difference learning y en qué se diferencia de los métodos tradicionales de aprendizaje supervisado? Explique el concepto de "error de diferencia temporal" y su papel en los algoritmos de aprendizaje por refuerzo**

Temporal Difference Learning es un método de aprendizaje por refuerzo que actualiza los valores de una política de aprendizaje sin tener que conocer las recompensas que se pueden obtener en el futuro. La mayor diferencia con el aprendizaje superbisado es que este no necesita de datos etiquetados. El "error de diferencia temporal" mide la diferencia entre la recompensa esperada y la obtenida, dejandonos hacerle cambios al aprendizaje del agente que estemos tratando de configurar.


**2. En el contexto de los juegos simultáneos, ¿cómo toman decisiones los jugadores sin conocer las acciones de sus oponentes? De un ejemplo de un escenario del mundo real que pueda modelarse como un juego simultáneo y discuta las estrategias que los jugadores podrían emplear en tal situación**

En juegos simultáneos, normalmente o lo esperado es que los jugadores tomen decisiones sin conocer las acciones de sus oponentes basándose en las mejores estrategias posibles, como pensando en que el jugador rival va a tomar ciertas decisiones. Un ejemplo es la subasta de sellos en eBay, en este, cada comprador hace su oferta sin saber la de los demás. Estrategias como la maximin (maximizar la peor ganancia posible) o la teoría de juegos bayesiana pueden ayudar a optimizar decisiones.


**3. ¿Qué distingue los juegos de suma cero de los juegos de suma cero y cómo afecta esta diferencia al proceso de toma de decisiones de los jugadores? Proporcione al menos un ejemplo de juegos que entren en la categoría de juegos de no suma cero y discuta las consideraciones estratégicas únicas involucradas**

Los juegos de suma cero tiene como principal detalle que la ganancia de un jugador es la pérdida del otro (por ejemplo en el ajedrez), mientras que en juegos de suma no cero, ambos pueden ganar o perder juntos (como por ejemplo, el dilema del prisionero). En suma cero, las estrategias suelen ser más competitivas ya que siempre se busca terminar ganandole al jugador contrario, mientras que en suma no cero, pueden surgir estrategias cooperativas ya que al ganar o perder juntos, puede que la mejor estrategia sea cooperar para ganar.

**4. ¿Cómo se aplica el concepto de equilibrio de Nash a los juegos simultáneos? Explicar cómo el equilibrio de Nash representa una solución estable en la que ningún jugador tiene un incentivo para desviarse unilateralmente de la estrategia elegida**

El equilibrio de Nash en juegos simultáneos ocurre cuando ningún jugador tiene un incentivo suficiente para cambiar su estrategia si los demás no lo hacen. Por ejemplo, en piedra-papel-tijera, si los dos jugadores tienen la opcion de elejir aleatoriamente con el mismo porcentaje de probabilidad, nadie puede mejorar de un solo lado evidentemente. Su aplicación hace que se deban considerar estrategias mixtas o combinadas y tener en cuenta como es que se comportan las personas.

Temporal Difference Learning se usa en entornos dinámicos para optimizar decisiones en tiempo real, como en robots o mercados financieros.
